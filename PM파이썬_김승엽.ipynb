```python
price = int(input("상품 가격(원): "))   # 입력: 숫자를 받아 price에 저장
total = price * 1.1                    # 처리: 부가세 10%를 더한 금액 계산
print("결제 금액:", int(total), "원")   # 출력: 소수점을 버리고 화면에 표시
print("결제 금액 3배: ", int(total) * 3, "원")
```

```python
onion = int(input("양파 가격(원): "))
tomato = int(input("토마토 가격(원): "))
potato = int(input("감자 가격(원): "))


onion = onion * 1.1
tomato = tomato * 1.1
potato = potato * 1.1

print("-----menu-------")
print("onion: ", int(onion), "원")
print("tomato: ", int(tomato), "원")
print("potato: ", int(potato), "원 ")
print("* 부가세 포함 금액입니다.")
print("----------------")

cs_onion = int(input("양파 몇 개 구매하시겠습니까?: "))
cs_tomato = int(input("토마토 몇 개 구매하시겠습니까?: "))
cs_potato = int(input("감자 몇 개 구매하시겠습니까?: "))

print("-----bill------")
print("onion: ", int(onion * cs_onion), "원")
print("tomato: ", int(tomato * cs_tomato), "원")
print("potato: ", int(potato * cs_potato), "원 ")
print("total: ", int(onion * cs_onion + tomato * cs_tomato + potato * cs_potato), "원")
print("* 부가세 포함 금액입니다.")
print("----------------")
```
